````python
# 함수만들기
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

def collect_list(keyword, page=1):
    'keyword로 다음 검색한 결과(해당 page)를 return'
    url = f'https://search.daum.net/search?w=news&nil_search=btn&DA=PGD&enc=utf8&cluster=y'
    params = {'q':keyword,
          'p':page}
    response = requests.get(url, params=params)
    soup = BeautifulSoup(response.text, # response.content
                    "html.parser")
    items_find_list = [] # 검색결과를 담을 dict list
    items_el = soup.select('div.item-title > strong.tit-g.clamp-g > a')
    for idx, item in enumerate(items_el):
        items_find_list.append({'no':(page-1)*10 + idx,
                           'title':item.text,
                           'link':item.attrs['href']})
    return items_find_list
````

### 4) User-Agent를 추가하여 크롤링
- urlopen()함수를 사용하면 크롤링이 안되는 사이트
- User-Agent를 추가하여 크롤링

````python
# 방법2 : urlopen()함수를 사용하면 크롤링이 안되는 사이트
# robots.txt에 Disallow로 되어 있는 경우
from urllib.request import urlopen, Request
import urllib.parse
word = '비트코인'
word = urllib.parse.quote(word)
print(word)
url = 'https://search.daum.net/search?w=news&nil_search=btn&DA=NTB&enc=utf8&cluster=y&cluster_page=1&q=\
%EB%B9%84%ED%8A%B8%EC%BD%94%EC%9D%B8'+word
# User-Agent를 추가하여 브라우저처럼 보이게 포장
headers = {'user-agent':'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/136.0.0.0 Safari/537.36'}
request = Request(url, headers=headers)
response = urlopen(request)
soup = BeautifulSoup(response, # response.content
                    "html.parser")

def collect_list(keyword, page=1):
    'keyword로 다음 검색한 결과(해당 page)를 return'
    url = f'https://search.daum.net/search?w=news&nil_search=btn&DA=PGD&enc=utf8&cluster=y'
    params = {'q':keyword,
          'p':page}
    response = requests.get(url, params=params)
    soup = BeautifulSoup(response.text, # response.content
                    "html.parser")
    items_find_list = [] # 검색결과를 담을 dict list
    items_el = soup.select('div.item-title > strong.tit-g.clamp-g > a')
    for idx, item in enumerate(items_el):
        items_find_list.append({'no':(page-1)*10 + idx,
                           'title':item.text,
                           'link':item.attrs['href']})
    return items_find_list

collect_list("청바지", 1)
````

````python
# 함수(keyword를 page수 만큼 검색한 결과를 데이터 프레임에 할당)
def get_items_data(keyword, pages):
    import requests
    from bs4 import BeautifulSoup
    import pandas as pd

    items_list = [] #크롤링한 데이터를 담을 list(2차원 list)

    for page in range(1, pages+1):
        url = f'https://kin.naver.com/search/list.naver?query={keyword}&page={page}'
    #     해당 url의 soup객체 -> 글제목, link를 가져와서 items_list에 append
        response = requests.get(url)
        soup = BeautifulSoup(response.text, 'html.parser')
        dts = soup.find_all('dt')

        for dt in dts:
            item = dt.find('a')
            items_list.append([item.text, item.attrs.get('href')])
            
    return pd.DataFrame(items_list, columns=['title', 'link'])
````

````python
def save_img(attr, idx, query, url):
    import requests
    '{attr}_{idx}_{query}.{확장자} 이미지 저장'
    file_extension = url.split('.')[-1]
    quote_index = file_extension.find('?')
    
    if quote_index != -1:
        file_extension = file_extension[:quote_index]
    print(file_extension)
    img = requests.get(url).content
    with open(f'ch14_image/{attr}_{idx}_{query}.{file_extension}', 'wb') as f:
        f.write(img)
````

#### # word image 100개 검색한 데이터는 csv파일로, 이미지는 ch14_image폴더에 저장되는 함수

````python
# word image 100개 검색한 데이터는 csv파일로, 이미지는 ch14_image폴더에 저장되는 함수
def get_naver_save_image(word):
    import requests
    import pandas as pd
    from dotenv import load_dotenv
    import os
    load_dotenv() 
    client_id = os.getenv('Client_ID')
    client_secret = os.getenv('Client_Secret')

    url = f'https://openapi.naver.com/v1/search/image?query={word}&display=100'
    headers = {'X-Naver-Client-Id': client_id,
               'X-Naver-Client-Secret':client_secret}
    response = requests.get(url, headers=headers)

    items = response.json()['items'] # 방법1에서만 가능한 함수
    items_list = []
    for idx, item in enumerate(items):
        title = item.get('title')
        link = item.get('link')
        thumbnail = item.get('thumbnail')
        sizeheight = item.get('sizeheight')
        sizewidth = item.get('sizewidth')
        items_list.append({
            'no': idx+1,
            'title': title,
            'link' : link,
            'thumbnail' : thumbnail,
            'sizeheight':sizeheight,
            'sizewidth' :sizewidth
        })
        # link와 thumbnail을 저장
        save_img('메인', idx+1, word, link)
        save_img('썸네일', idx+1, word, thumbnail)
        # 20%, 40%, 60%, 80% 지점에서 message 출력
        if idx/20 == round(idx/20):
            print(f'= = = {idx} % 진행 중입니다 = = =')       
    df = pd.DataFrame(items_list)
    df.to_csv(f'ch14_image/{word}.csv', index=False, encoding='cp949')
    print('~ ~ ~ 이미지 및 csv 파일 저장 완료 ~ ~ ~')
````

````python
import requests
from bs4 import BeautifulSoup
import pandas as pd

size = 48
pages = 1
items_list = []

for page in range(1, pages+1):
    url = f'https://www.yes24.com/product/category/bestseller?categoryNumber=001&pageNumber={page}&pageSize={size}'
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    rank = soup.select('em.ico.rank')
    gd_name = soup.select('a.gd_name')    
    writer = soup.select('span.authPub.info_auth > a')
    price = soup.select('strong.txt_num > em.yes_b')
    company = soup.select('span.authPub.info_pub > a')

    for i in range(len(gd_name)):
        items_list.append({
            '순위': rank[i].text,
            '책이름': gd_name[i].text,
            '저자': writer[i].text,
            '출판사': company[i].text,
            '가격': price[i].text+'원',
            })

df = pd.DataFrame(items_list, columns=['순위', '책이름', '저자', '출판사', '가격',])
df
````
