# 1. 연관분석 개요
- 데이터들 사이에 자주 발생하는 속성을 찾고, 그 속성들 사이에 연관성이 어느 정도 있는지 분석
- 활용분야 : 이벤트 미리 감지(사기적발..), 신상품카테고리 구성

- [조건:left-hand side:오렌지주스]->[결과:right-hand side:와인]
- 연관분석과 관련된 지표
- 1. 지지도(support) : 얼마나 자주 함께 나타나는지
    - (left hand size와 right hand size)의 항목수/전체항목수 = 0.2
- 2. 신뢰도(confidence) : 조건이 오면 얼마나 자주 나타나는지
    - (lhs->rhs)의 데이터 항목수/전체항목수 = 01/2 = 0.5
- 3. 향상도(lift) : 우연히 발생한 규칙은 아닌지 확인
    - (lhs->rhs)의 데이터 항목수/((lhs)의 지지도*(rhs)의 지지도)
    - => 0.2 / (0.4*0.6) = 0.2/0.24 = 0.833
    - 향상도 < 1 : 기대가 낮다
    - 향상도 > 1 : 기대가 높다(있다)

# 2. 연관분석 구현

```python
import csv
transaction = []
with open('data/cf_basket.csv', 'r', encoding='utf-8') as f:
    csvdata = csv.reader(f)
#     print(list(csvdata))
    transaction = list(csvdata)
transaction
```

```python
from apyori import apriori
rules = apriori(transaction, # 2차원데이터
               min_support = 0.15,
               min_confidence = 0.1,
               )
rules = list(rules)
len(list(rules))
```

```python
rules[10]
```

```python
rule = rules[10]
support = rule[1]
order_st = rule[2]
for item in order_st:
    lhs = item[0]
    rhs = item[1]
    confidence = item[2]
    lift = item[3]
    if lift > 1:
        print("{}=>{}\t {}\t {}\t {}".format(lhs, rhs, support, round(confidence,2), round(lift,2)))
```

# 3. 경주/전주 여행 자료 연관분석

```python
import pandas as pd
from konlpy.tag import Hannanum, Komoran
df = pd.read_csv('data/naver_kin.csv', sep='\t')
total_text_list = df['total_text'].to_list()
# total_text_list[:2]

analyzer = Komoran()
total_noun_list = []
select_pos = ['NC','NQ'] # 보통명사, 고유명사 (Hannanum)
select_pos = ['NNP','NNG'] # Kkma, Komoran 보통명사, 고유명사
불용어 = {'여행'}
for total_text in total_text_list:
#     total_noun = analyzer.nouns(total_text) # 결과물이 좋지 않아서
    total_noun = [token for token, tag in analyzer.pos(total_text) #, ntags=22(한나눔할때는)
                  if tag in select_pos and token not in 불용어 and len(token)>1]
    total_noun_list.append(total_noun)
print(total_noun_list[:2])
```

```python
%%time
rules = apriori(total_noun_list, min_support=0.15, min_confidence=0.3)
rules = list(rules)
len(rules)
```

```python
# rules_df.loc[0] = ['와인', '오렌지', 0.15, 0.5, 1.1] # 식으로 for문내에서 데이터추가
idx = 0
for rule in rules:
    support = rule[1]
    order_st = rule[2]
    for item in order_st:
        lhs = ', '.join([data for data in item[0]])
        rhs = ', '.join([data for data in item[1]])
        confidence = item[2]
        lift = item[3]
        if lift > 1:
            rules_df.loc[idx] = [lhs, rhs, support, round(confidence,2), round(lift,2)]
            idx += 1
            
rules_df.sort_values(by=['향상도','신뢰도'], ascending=False, inplace=True)

# rules_df = rules_df.reset_index(drop=True)
# rules_df.index = rules_df.index + 1
#두줄을 한줄로하면
# rules_df.index = range(1, len(rules_df) + 1)
rules_df = rules_df.reset_index(drop=True)
```
