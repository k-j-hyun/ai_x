# 1. 문맥을 이용하여 모델만들기

## 1. 데이터 전처리

```python
text = """경마장에 있는 말이 뛰고 있다
그의 말이 법이다
가는 말이 고와야 오는 말이 곱다"""
# text1 = "장마철이 오고 있다"
```

```python
# from keras_preprocessing.text import Tokenizer
# t = Tokenizer()
# # 학습시킬 모델을 리스트 안에 넣음 text -> 숫자로 변환 함수 .fit_on_texts([])
# t.fit_on_texts([text, text1])
# encoded = t.texts_to_sequences([text, text1])
# print(encoded)
# print(t.word_index)
```

```python
from keras_preprocessing.text import Tokenizer
t = Tokenizer()
# 학습시킬 모델을 리스트 안에 넣음 text -> 숫자로 변환 함수 .fit_on_texts([])
t.fit_on_texts([text])
encoded = t.texts_to_sequences([text])
print(encoded)
print(t.word_index)
```

```python
# 문자열 리스트를 인덱스 시퀀스로 변환
print(t.texts_to_sequences(['경마장에 말이 있다','말이 뛴다']))
print(t.texts_to_sequences(['가는 말이 곱다'])[0])
```

```python
text = """경마장에 있는 말이 뛰고 있다
그의 말이 법이다
가는 말이 고와야 오는 말이 곱다"""
```

```python
for key, value in t.word_index.items():
    print(key, value)
```

```python
# 학습을 시키기 위해 ['경마장에 있는','경마장에 있는 말이',...]
sequences = []
for line in text.split('\n'):
    print('원문장 :', line)
    encoded = t.texts_to_sequences([line])[0] # [0]을 붙혀야 리스트로 만들어짐
    print('encoded된 문장 :', encoded)
    for i in range(0, len(encoded)-1): # i:시작index
        for j in range(i+2, len(encoded)+1): # j:끝나는 index바로 뒤
            sequences.append(encoded[i:j])
print('sequences와 해석 출력')
for sequence in sequences:
    for word_seq in sequence:
        for word, id in t.word_index.items():
            if word_seq==id:
                print("{}:{}".format(word_seq, word), end=' ')
                break
    print()
```

```python
print([len(sequence) for sequence in sequences])
my_len = max(len(sequence) for sequence in sequences)
my_len
```

```python
# sequences를 훈련 가능하도록 모두 6개로 조정
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_sequences = pad_sequences(sequences=sequences,
                                maxlen=my_len,
                                padding='pre', # pre=앞에0, post=뒤에0을 붙히는것
                                )
# padded_sequences
padded_sequences.shape, type(padded_sequences)
```

```python
# 독립변수(X)와 타겟변수(y)로 분리
X = padded_sequences[:, :-1] # 모든행에 맨마지막 바로 앞까지 슬라이싱
y = padded_sequences[:, -1]
y
```

```python
# X를 엠베딩 작업할 때 필요한 입력 bit 수 (단어수 + 1)
# 컴퓨터가 임베딩 작업할 때 필요한 입력 bit 수(단어 갯수) (단어수 + 1)
input_dim = len(t.word_index)+1
input_dim
```

```python
# y의 원핫인코딩
from tensorflow.keras.utils import to_categorical
Y = to_categorical(y, input_dim)
X.shape, Y.shape, input_dim
```

## 2. 모델 생성하기

```python
# 2. 모델 생성하기
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding # X(i)의 원핫인코딩한 희소행렬 변환 i=1~5 이런식의 입력값
from tensorflow.keras.layers import SimpleRNN, Dense 
import matplotlib.pyplot as plt
```

## RNN, 3. 모델학습과정설정, 4. 모델 학습시키기

```python
model = Sequential()

model.add(Embedding(input_dim=input_dim, # 12개를 입력했으니 자동으로 원핫인코딩해줌
                    output_dim=10, # 임베딩 출력의 갯수
                    input_length=X.shape[1], # 입력 단어(X)의 수
                    ))

# RNN
model.add(SimpleRNN(32)) # 입력층
model.add(Dense(12, activation='softmax'))
print(model.summary())

# 3. 모델 학습 과정 설정
model.compile(loss='categorical_crossentropy',
             optimizer='adam',
             metrics=['accuracy'])

# 4. 모델 학습시키기
hist = model.fit(X, Y, epochs=300, verbose=2)
```

## 5. 모델 학습과정 시각화

```python
# 5. 모델 학습과정 시각화
fig, loss_ax = plt.subplots(figsize=(12,6))
loss_ax.plot(hist.history['loss'], 'y', label='train loss')

acc_ax = loss_ax.twinx()
acc_ax.plot(hist.history['accuracy'], 'g', label='train accuracy')

loss_ax.set_xlabel('epochs')
loss_ax.set_ylabel('loss')
acc_ax.set_ylabel('accuracy')
loss_ax.legend(loc='center right')
acc_ax.legend()
plt.show()
```

```python
# 모델 사용하기 (경마장에 있는 말이 -> 뛰고)
input_word = input('입력단어 :')
encoded = t.texts_to_sequences([input_word])[0]
input_data = pad_sequences([encoded], maxlen=my_len-1, padding='pre')
print('입력데이터 :', input_data)
result = model.predict(input_data).argmax()
word = t.index_word[result]
print('예측된 단어 :', word)
```

# 2. 다음 문맥 예측해 보기

```python
# "경마장에" 이후에 올 단어를 n개 예측하기 ex) 3개 예측 -> 경마장에 있는 말이 뛰고
# 웹에서는 def sentence_generation(model, t, current_word, n) 다 넣어야함
def sentence_generation(current_word, n): # t=tokenizer 숫자=단어 교환
#     print('입력된 단어 :', current_word)
    for i in range(1, n+1):
        encoded = t.texts_to_sequences([current_word])[0]
        input_data = pad_sequences([encoded], maxlen=my_len-1, padding='pre')
        result = model.predict(input_data, verbose=0).argmax()
        word = t.index_word[result]
        current_word = current_word + ' ' + word
    return current_word
#         print("{}번째 생성 문맥 {} ({}{}) : ".format(i,
#                                                current_word,
#                                                encoded[0],
#                                                result))
sentence_generation('경마장에', 3)
```
