```python
from qdrant_client import QdrantClient

API_KEY = "네_클라우드_API_KEY를_여기에_넣으세요"
ENDPOINT = "네_클라우드_엔드포인트.qdrant.cloud"  # .qdrant.cloud 포함해서 정확히 복사한 주소

client = QdrantClient(
    url=f"https://{ENDPOINT}",
    api_key=API_KEY,
    prefer_grpc=False  # Cloud에서는 기본적으로 HTTP 사용
)
```

# 컬렉션 생성 벡터 예시

```python
from qdrant_client.models import VectorParams, Distance, PointStruct

# 컬렉션 생성 (벡터 차원, 거리 측정 방식 설정)
client.recreate_collection(
    collection_name="my_cloud_collection",
    vectors_config=VectorParams(size=128, distance=Distance.COSINE)
)

# 벡터 예시 2개 업로드
points = [
    PointStruct(id=1, vector=[0.1]*128, payload={"text": "클라우드 첫 문서"}),
    PointStruct(id=2, vector=[0.2]*128, payload={"text": "클라우드 두번째 문서"}),
]
client.upsert(collection_name="my_cloud_collection", points=points)

# 유사 벡터 검색
query_vector = [0.1]*128
results = client.search(collection_name="my_cloud_collection", query_vector=query_vector, limit=2)

for res in results:
    print(f"id: {res.id}, score: {res.score:.4f}, text: {res.payload['text']}")
```
