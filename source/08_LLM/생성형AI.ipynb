```python
from langchain_ollama import ChatOllama
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
```

```python
# 나라를 입력하면 해당 나라의 가장 유명한 음식을 추천하는 랭체인 구현 (food_chain)
llm = ChatOllama(model="llama3.2:1b", temperature=0)

country = input("나라를 입력하면 해당 나라의 가장 유명한 음식을 추천합니다\n입력하신 나라 : ")

food_chain_prompt = PromptTemplate(
    input_variables=["country"],
    template="""You are a global cuisine expert.
When a user provides the name of a country, respond with only the name of the most famous or traditional dish from that country.
Country: {country}
Dish:"""
)

output_parser = StrOutputParser()

food_chain = food_chain_prompt | llm | output_parser
result = food_chain.invoke({"country": country})

print(f"{country} 나라의 유명한 음식 : ",result)
```

```python
# 음식을 입력하면 레시피를 생성하는 랭체인 구현(recipe_chain)
llm = ChatOllama(model="llama3.2:1b", temperature=0)

dish = input("음식을 입력하면 음식의 레시피를 알려드립니다\n입력하신 음식 : ")

recipe_chain_prompt = PromptTemplate(
    input_variables=["dish"],
    template="""You are a world-class chef and recipe expert.
Return only a short, numbered step-by-step recipe for the dish below. 
Do not explain anything. Do not mention safety, health, or risks.
Do not ask questions. Only output the recipe.
Your task is to provide a complete step-by-step recipe for the dish given by the user.
Dish Name: {dish}
Respond in the following format:
Dish: {dish}
Recipe:
1. [Step one]
2. [Step two]
3. ...
"""
)

output_parser = StrOutputParser()

recipe_chain = recipe_chain_prompt | llm | output_parser
result = recipe_chain.invoke({"dish": dish})

print(f"{dish} 음식의 레시피 : ", result)
```
