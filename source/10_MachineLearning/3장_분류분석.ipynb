## 2-1 확률적 모형
    - 주어진 데이터에 각 클래스가 정답일 확률을 계산하는 모형
    - predict_proba() : 각 분류 클래스별 확률을 알려줌(일부의 판별함수 모형에 있긴 함)

### 1) 확률적 생성 모형

    - ex. Quadratic Discriminant Analysis, 나이브 베이지안(NaiveBayes)
    
### 2) 확률적 판별 모형
    - 확률을 계산하는 함수 이용
    - ex. 의사결정나무, 로지스틱 회귀
    
## 2-2 판별함수 모형

    - decision_function()
    - ex. 퍼셉트론, 서포트벡터머신, MLP(신경망)

```python
# 샘플 데이터
from sklearn.datasets import make_classification
X, y = make_classification(n_features=2, # 독립변수 갯수
                          n_informative=2, # 타겟변수에 영향을 주는 독립변수 갯수
                          n_redundant=0, # 타겟 변수에 영향을 안 주는 독립변수 갯수
                          n_clusters_per_class=1, # 클래스 내 서브 그룹수
                          n_classes=2, # 타겟변수 class 수(군의 수)
                          n_samples=100, # 기본값 100 , 데이터 갯수(행수)
                          random_state=9,
                          )
X.shape, y.shape
```

#### 분류 모형의 시각화 decision_tree

```python
# 분류 모형의 시각화
```

```python
import os
import io
from sklearn.tree import export_graphviz
import pydot
from IPython.core.display import Image
```

```python
def draw_decision_tree(model, feature_names):
    dot_buf = io.StringIO()
    export_graphviz(model, out_file=dot_buf, feature_names=feature_names)
    graph = pydot.graph_from_dot_data(dot_buf.getvalue())[0]
    image = graph.create_png()
    return Image(image)
draw_decision_tree(dt_model, feature_names=feature_names)
```

```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
```

```python
def plot_decision_regions(X, y, model, title):
    species = ['setosa', 'versicolor', 'virginica']
    resolution = 0.01
    markers = ('s', '^', 'o')
    colors = ('red', 'blue', 'lightgreen')
    cmap = mpl.colors.ListedColormap(colors)
    
    x1_min, x1_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    x2_min, x2_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx1, xx2 = np.meshgrid(np.arange(x1_min, x1_max, resolution),
                np.arange(x2_min, x2_max, resolution))
    Z = model.predict(
        np.array([xx1.ravel(), xx2.ravel()]).T).reshape(xx1.shape)
    
    plt.contour(xx1, xx2, Z, cmap=mpl.colors.ListedColormap(['k']))
    plt.contourf(xx1, xx2, Z, alpha=0.4, cmap=cmap)
    plt.xlim(xx1.min(), xx1.max())
    plt.ylim(xx2.min(), xx2.max())
    
    for idx, cl in enumerate(np.unique(y)):
        plt.scatter(x=X[y == cl, 0], y=X[y == cl, 1], alpha=0.8,
            c=[cmap(idx)], marker=markers[idx], s=80, label=species[cl])
    
    plt.xlabel(data.feature_names[2])
    plt.ylabel(data.feature_names[3])
    plt.legend(loc='upper left')
    plt.title(title)
    
    return Z
```

```python
draw_decision_tree(dt_model, feature_names=data.feature_names[2:])
```

```python
plot_decision_regions(X, y, dt_model, "Depth 1")
plt.show()
```

```python
dt_model5 = DecisionTreeClassifier(criterion="entropy",
                                  max_depth=5,
                                  random_state=0)
dt_model5.fit(X, y)
draw_decision_tree(dt_model5, feature_names=data.feature_names[2:])
```

```python
plt.figure(figsize=(10,6))
plot_decision_regions(X, y, dt_model5, "Depth 5")
plt.show()
```

## 6-3. ROC커브를 이용한 성능 비교
- ROC그래프의 가로축 위양성율(fprate : False Positive rate), 세로축 recall값

```python
from sklearn.datasets import make_classification
X, y = make_classification(weights=[0.95,0.05], # negative, positive 비율 기본값 0.5:0.5
                          n_samples=1000,
                          random_state=5)
X.shape, y.shape
```

```python
np.unique(y), y.mean()
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(random_state=0)
model.fit(X, y)
```

```python
from sklearn.metrics import roc_curve
fpr, tpr, thr = roc_curve(y, model.decision_function(X))
plt.figure(figsize=(2,2))
plt.plot(fpr, tpr)
plt.xlabel('false positive rate') # 위양성율
plt.ylabel('recall')
plt.show()
```

```python
y_proba = model.predict_proba(X)[:,1]
fpr1, tpr1, thr1 = roc_curve(y, y_proba)

from sklearn.metrics import roc_curve
fpr, tpr, thr = roc_curve(y, model.decision_function(X))
plt.figure(figsize=(2,2))
plt.plot(fpr1, tpr1)
plt.xlabel('false positive rate') # 위양성율
plt.ylabel('recall')
plt.show()
```

## 6-4 두 모형의 혼동행렬이 같은 경우(roc_curve사용)

```python
from sklearn.datasets import make_classification
X, y = make_classification(weights=[0.95,0.05], # negative, positive 비율 기본값 0.5:0.5
                          n_samples=1000,
                          random_state=5)
X.shape, y.shape
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

model1 = LogisticRegression().fit(X, y)
model2 = SVC(gamma=0.0001, # 값이 작을수록 경계면이 부드러운 곡선
            C=3000, # 오차 허용 범위 : 과적합 신경안쓰고 학습 (크면클수록)
            probability=True, # true를 쓰면 predict_proba가 가능, false면 안됨
            ).fit(X, y)
```

```python
y_hat1 = model1.predict(X)
y_hat2 = model2.predict(X)
```

```python
pd.crosstab(y, y_hat1)
```

```python
pd.crosstab(y, y_hat2)
```

```python
# 혼동 행렬이 같으므로 모든 score 값이 같음
# roc 커브로 두 모델 성능 비교
```

```python
from sklearn.metrics import roc_curve
fpr1, tpr1, thr1 = roc_curve(y, model1.decision_function(X))
fpr2, tpr2, thr2 = roc_curve(y, model2.decision_function(X))
```

```python
plt.plot(fpr1, tpr1, color='r')
plt.plot(fpr2, tpr2, color='b')
plt.show()
```

```python
# AUC(Area Under Curve) : roc커브 아랫부분의 영역
# AUC = 1 : 완벽한 모델
# AUC < 0.5 : 최악의 모델
# AUC = 0.5 : 동전 던지기와 일치
from sklearn.metrics import auc
auc(fpr1, tpr1), auc(fpr2, tpr2)
```
